### Impute data for the predictors using IterativeImputer
For the rest of the numerical predictors, we will be using sklearn's IterativeImputer to impute missing predictors

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

Create a list with all 5 numerical predictors 

```python
numerical_predictors = [
    'rate_unemployment',
    'percent_college',
    'percent_married',
    'median_income',
    'percent_lunch'
]
```

We will create two list with minimum and maximum bounds such that imputer does not impute unrealistic data. For median income, since it is not measured in percentage (0 - 1) scale, we set the max bound to np.inf (infinite).

```python
min_value_list = [0,0,0,0,0] # setting minimum bounds for each of the numerical predictors
max_value_list = [1,1,1,np.inf,1] # setting max bounds for each of the numerical predictors. 
```

```python
imputer = IterativeImputer( min_value= min_value_list, max_value=max_value_list) # models the variable with missing values as a function of other variables
```

```python
df_imputed = df_clean.copy()
```

```python
imputer.fit(df_imputed.loc[:, numerical_predictors]) # learn the relationships between the columns
```

```python
df_clean.info()
```

```python
df_imputed.loc[:,numerical_predictors] = imputer.transform(df_imputed.loc[:, numerical_predictors])
```

```python
df_imputed.info()
```

```python
df_imputed[numerical_predictors].agg(['min', 'max'])
```

## Data Preparation Summary
The data preparation process involved importing, investigating, cleaning, and merging the EdGap and school information datasets.
Relevant variables were selected and renamed using consistent naming conventions. The datasets were merged using the school id, and missing values and potentially invalid numerical observations were investigated.
Three invalid ACT scores and 20 negative free/reduced-price lunch participation values were identified. The dataset was then filtered to include schools offering grade 12.
Observations with missing ACT scores were removed, while missing numerical predictor values were imputed using `IterativeImputer`, with predefined minimum and maximum bounds to maintain data validity.
The resulting dataset contains 7,857 schools and 14 numerical and categorical variables, with no missing values in the selected analytical variables. The prepared data were exported as `education_final.csv` for exploratory data analysis and regression modeling.
