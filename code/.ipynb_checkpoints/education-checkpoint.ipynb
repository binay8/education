### Rename Columns

```python
edgap = edgap.rename(
    columns={'NCESSCH School ID': 'id', 
             'CT Pct Adults with College Degree': 'percent_college' ,
             'CT Unemployment Rate': 'rate_unemployment' ,
             'CT Pct Childre In Married Couple Family': 'percent_married', 
             'CT Median Household Income':'median_income',
             'School ACT average (or equivalent if SAT score)':'average_act',
             'School Pct Free and Reduced Lunch':'percent_lunch',
            })
```

### Address incorrect data using df.loc[row_condition, column_name]= new_val
We will address incorrect data by overriding existing values with np.nan

```python
df.loc[df['average_act'] < 1, 'average_act' ] = np.nan
```

```python
df.loc[df['percent_lunch'] <0, 'percent_lunch'] = np.nan
```

Verify min max to see if the values are correct

```python
df.select_dtypes(include=['number']).agg(['min', 'max']).round(2)
```

### Impute data for the predictors using IterativeImputer

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
numerical_predictors = [
    'rate_unemployment',
    'percent_college',
    'percent_married',
    'median_income',
    'percent_lunch'
]
```

```python
min_value_list = [0,0,0,0,0] # setting minimum bounds for each of the numerical predictors
max_value_list = [1,1,1,np.inf,1] # setting max bounds for each of the numerical predictors
```

```python
imputer = IterativeImputer( min_value= min_value_list, max_value=max_value_list) # models the variable with missing values as a function of other variables
```

```python
df_imputed = df_clean.copy()
```

```python
imputer.fit(df_imputed.loc[:, numerical_predictors]) # learn the relationships between the columns
```

```python
df_clean.info()
```

```python
df_imputed.loc[:,numerical_predictors] = imputer.transform(df_imputed.loc[:, numerical_predictors])
```

```python
df_imputed.info()
```

```python
df_imputed[numerical_predictors].agg(['min', 'max'])
```

```python
df_imputed[numerical_predictors].isna().sum()
```
